## 1 Pseudokode

En **algoritme** er en oppskrift som beskriver hvordan vi løser et problem. Den kan utføres av en datamaskin eller for hånd. Et **dataprogram** er instruksjoner datamaskinen kan forstå, og kan inneholde én eller flere algoritmer. Algoritmen er abstrakt, programmet er en konkret realisering av den.

Programmeringsspråk har strenge regler som gjør koden vanskelig å lese for mennesker. **Pseudokode** er en generell beskrivelse av en algoritme, skrevet på et mer naturlig språk. Den gjør det lettere å kommunisere og samarbeide, og er nyttig når man lærer å programmere.

Pseudokode er ikke standardisert, men vi bruker grunnelementer som ligner det vi kjenner:

| Element | Pseudokode |
|---|---|
| Inndata | `LES tekst` |
| Utdata | `SKRIV "Hei!"` |
| Beregning | `total = antall * pris` |
| Valg | `HVIS tall > 10 SÅ … ELLERS …` |
| Løkke | `GJENTA 10 GANGER: …` |

Vi trenger ingen spesielle verktøy – en teksteditor eller en tavle holder.

### Eksempel: finn det største tallet

Algoritmen med ord:

1. Sett maks til det første tallet i listen.
2. Gå gjennom hvert tall i listen fra og med det andre.
3. Hvis tallet er større enn maks, sett maks til tallet.
4. Når alle tallene er sjekket, inneholder maks det største tallet.
5. Skriv ut maks.

Som pseudokode:

```
LES liste
SETT maks til første tall i liste
GJENTA for hvert tall i liste:
    HVIS tall > maks:
        SETT maks til tall
SKRIV maks
```

En stor fordel med pseudokode er at den er lett å oversette til et hvilket som helst programmeringsspråk:

````python
liste = [3, 7, 1, 9, 4, 2, 8, 6, 5]
maks = liste[0]

for tall in liste:
    if tall > maks:
        maks = tall

print(maks)
````

Det samme i JavaScript:

```javascript
let liste = [3, 7, 1, 9, 4, 2, 8, 6, 5];
let maks = liste[0];
for (let i = 1; i < liste.length; i++) {
    let tall = liste[i];
    if (tall > maks) {
        maks = tall;
    }
}
console.log(maks);
```

## 2 Flytskjema

**Flytskjemaer** (flytdiagrammer) viser kontrollflyten i et program grafisk. De hjelper oss å planlegge strukturert og er et alternativ for dem som foretrekker det visuelle.

| Symbol | Form | Brukes til |
|---|---|---|
| Terminal | avrundet (idrettsbane) | start og stopp |
| Parallellogram | skrått | inn- og utdata |
| Rektangel | firkant | beregninger |
| Diamant | rombe | valg (test/sjekk) |
| Pil | → | flyten mellom symbolene |
| Tekstboble | | kommentarer |

Løkker lager vi selv med en kombinasjon av symbolene. Flytskjemaet for å finne det største tallet (med en `for`-løkke):

```mermaid
flowchart TD
    S([Start]) --> L[/Les liste/]
    L --> B["maks = liste[0]<br>N = antall<br>i = 1"]
    B --> T{i < N ?}
    T -- Ja --> V{"liste[i] > maks ?"}
    V -- Ja --> M["maks = liste[i]"]
    V -- Nei --> I[i += 1]
    M --> I
    I --> T
    T -- Nei --> U[/Skriv maks/]
    U --> E([Stopp])
```

Verktøy: [draw.io](https://www.drawio.com/) (finnes som VS Code-utvidelse – lag en fil med filtypen `.drawio`), Lucidchart og Microsoft Visio. Diagrammet over er laget med Mermaid, som vises direkte i JupyterLab og VS Code.

Flytskjemaer kan være lettere å forstå for dem som ikke kan programmere. Pseudokode egner seg bedre for detaljerte og komplekse algoritmer som blir uoversiktlige som figurer.

## 3 Wireframes

**Wireframes** er skisser som viser hvordan et brukergrensesnitt skal se ut og virke. Pseudokode og flytskjema beskriver logikken; wireframes beskriver utseendet og funksjonaliteten for brukeren. Utviklere, designere, prosjektledere og brukere kan samarbeide og finne problemer tidlig – endringer senere, når mye kode er skrevet, tar mer tid og koster mer.

Selv enkle wireframes krever mange beslutninger: Hvilke enheter skal støttes? Hvordan håndteres skjermstørrelse og -retning? Hvilke tilbakemeldinger får brukeren? Radioknapper eller nedtrekksliste? Verktøytips? Derfor finnes egne stillinger for **UX/UI-designere** (*User Experience/User Interface*).

Wireframes kan tegnes for hånd på papir eller tavle – raskt og fint for å eksperimentere. Digitale verktøy gir flere muligheter og kan eksportere til PDF og bilder: [Figma](https://www.figma.com/), [Balsamiq](https://balsamiq.com/), Sketch, Adobe XD – og [Salt](https://plantuml.com/salt) i PlantUML. Vi anbefaler [Pencil](https://pencil.evolus.vn/), som er gratis og enkel.

Wireframen for treningskalkulatoren fra 2.1 i PlantUML Salt:

```
@startsalt
{
  ^Velg aktivitet^
  Velg intensitet:
  () Lav
  (X) Middels
  () Høy
  Oppgi varighet på trening i minutter
  "     "
  [Beregn kaloriforbruk]
  {+ Svar
     .
  }
}
@endsalt
```
