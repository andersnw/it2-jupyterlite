## 1 Bildemanipulering er ikke nytt

Bilder har blitt manipulert siden de første fotografiene på 1800-tallet. Et kjent eksempel er Sovjetunionen under Stalin, der personer som hadde falt i unåde, ble fjernet fra gruppebilder. Det ble enklere da John og Thomas Knoll laget Photoshop på slutten av 1980-tallet. Vi kan justere farger og lys, men også forvrenge virkeligheten: fjerne kviser, gjøre tenner hvitere, endre ansiktsform, gjøre folk slankere – eller fjerne dem helt. Kjendiser kan ikke hindre at det florerer av falske, pornografiske bilder av dem på nettet.

## 2 Hva er deepfakes?

Med kunstig intelligens kan vi nå manipulere **video og lyd** slik at vi kan «få hvem som helst til å si hva som helst». Det ser og høres ut som en person vi kjenner, men det er en annen som opptrer – ansiktet og stemmen er byttet ut. Fenomenet kalles **deepfakes**, og kvaliteten blir stadig bedre. En kjent video av Barack Obama ble laget allerede i 2018.

Er dette noe vi bør bekymre oss for, eller er det gammeldags desinformasjon med en ny vri?

## 3 Muligheter

* **Film:** Skuespillere kan gjøres yngre (*The Irishman*), erstattes av deepfakes av seg selv, eller byttes ut med digitale skuespillere. Scener kan endres uten å spilles inn på nytt, og dubbing blir mer realistisk.
* **Museum og skole:** Dalí-museet i Florida har «vekket» kunstneren til live. Historiske personer kan gjøre undervisningen mer levende.
* **Lydbøker og språk:** Kjente skuespillere kan «lese inn» lydbøker uten å lese dem selv, og influensere kan «snakke» språk de ikke kan.
* **Universell utforming** for personer med nedsatt funksjonsevne.
* **Kundeservice** med digitale personer.
* **Humor:** Komikere har fått et nytt verktøy, og apper som Reface og Dream lar hvem som helst more seg.

## 4 Utfordringer

Teknologi kan brukes og misbrukes – fra atomkraft og droner til smarttelefoner og sosiale medier.

* **Pornografi:** Det meste av deepfakes er porno der ansiktene til kvinner er byttet ut – og ikke bare kjendiser rammes. Det er trakassering og ulovlig, og åpner for utpressing.
* **Svindel:** Med falsk lyd ble et britisk selskap svindlet for over to millioner kroner i 2019. Nettdating-svindel har kostet folk milliarder.
* **Reklame og falske personer:** «Sokkedukker» kan opptre som nettroll, og produktanmeldelser kan forfalskes.
* **Demokrati:** Vi kan bli lurt til å tro at politikere har sagt ting de ikke har sagt. Sammen med alternative fakta og falske nyheter er dette en trussel mot fri presse og offentlig debatt.
* **Kunst eller fakta?** Dramaserien *Atlantic Crossing* (NRK, 2020) ble kritisert for historieforfalskning. Produsenter av deepfakes står overfor lignende dilemmaer.

## 6 Prøv selv: en enkel bildemanipulering

Deepfakes bruker avanserte nevrale nettverk, men prinsippet – å endre piksler – kan vi prøve med noen få linjer. Vi lager et bilde, «klipper ut» en del og limer den inn et annet sted. Et bilde er bare en tabell med tall.

```python
import numpy as np
import matplotlib.pyplot as plt

# Et enkelt «bilde»: himmel, sol og gress
bilde = np.zeros((100, 150, 3))
bilde[:, :] = [0.6, 0.8, 1.0]                     # himmel
bilde[70:, :] = [0.2, 0.6, 0.2]                   # gress
y, x = np.ogrid[:100, :150]
bilde[(x - 30)**2 + (y - 25)**2 < 150] = [1, 0.9, 0.2]   # sol

manipulert = bilde.copy()
manipulert[10:40, 100:130] = bilde[10:40, 15:45]  # kopier sola til et nytt sted

fig, akser = plt.subplots(1, 2, figsize=(9, 3))
for ax, b, tittel in zip(akser, [bilde, manipulert], ['Original', 'Manipulert: to soler']):
    ax.imshow(b); ax.set_title(tittel); ax.axis('off')
plt.show()
```
