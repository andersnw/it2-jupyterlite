## Innledende eksempel
"Programmer en rutine som beregner flytid når brukeren skriver inn avstand og fart. (Fart oppgis i knop, og avstand oppgis i kilometer, 1 knop = 1,852 km/t.) Angi svaret i timer og minutter eller som desimaltall."

Programmet under vil generere følgende feilmelding: `TypeError: can't multiply sequence by non-int of type 'float'`. For å forstå denne feilmeldingen, må vi kjenne til ulike **datatyper** i programmering.

```python
avstand_i_km = input('Oppgi avstand i km: ')
fart_i_knop = input('Oppgi fart i knop: ')
fart_i_kmpt = fart_i_knop * 1.852
flytid      = avstand_i_km / fart_i_kmpt
print(flytid)
```

## Grunnleggende datatyper
Grunnleggende datatyper inkluderer heltall, desimaltall, tekst og boolske verdier (sann eller usann).

I Python kan vi bruke `type`-funksjonen (BIF) til å sjekke datatype til en verdi eller variabel. De tilsvarende datatypene i Python er `int`, `float`, `str` og `bool`.

```python
print(type(123))     # <class 'int'>
print(type(1.23))    # <class 'float'>
print(type('Askim'))  # <class 'str'>
print(type(True))    # <class 'bool'>
b = 3.0
a = "2"
print(type(a))  # <class 'str'>
print(type(b))  # <class 'int'>
```

## IntelliSense i Visual Studio Code
Hvis vi holder musepekeren over `avstand_i_km` i programmet, vises informasjonen som indikerer at variabelen er av typen `str`. Disse hjelpemidlene *(code hints)* er en del av IntelliSense-funksjonen i Visual Studio Code.

Hvis vi holder vi musepekeren over `input`-funksjonen, vises informasjonen som indikerer at den returnerer en tekst (`str`).

Dette er årsaken til feilmeldingen: `input` returnerer en tekst, og det er ikke tillatt å multiplisere en tekst (sekvens av tegn) med et desimaltall. Hvis det hadde vært et heltall, ville det vært i orden.

```python
# 3.0 * 'Hurra! ' gir feilmelding: TypeError
# 3 * 'Hurra! '  fungerer fint:
print(3 * 'Hurra! ')  # 'Hurra! Hurra! Hurra! '
```

Hvis teksten representerer et tall, kan vi konvertere den til et desimaltall ved å bruke `float`-funksjonen (BIF).

Når vi skriver, gir IntelliSense oss forslag for automatisk fullføring, også kjent som *auto complete*. Vi kan velge alternativene med **piltastene** og fullføre med **Tab** eller **Enter**, eller vi kan fortsette å skrive selv. Vi kan også få tilgang til IntelliSense ved å bruke **Ctrl+mellomrom**-kombinasjonen.

Hvis vi nå gjør endringer i de to første linjene i programmet, vil det fungere bedre.

```python
avstand_i_km = float(input('Oppgi avstand i km: '))
fart_i_knop  = float(input('Oppgi fart i knop: '))
fart_i_kmpt = fart_i_knop * 1.852
flytid      = avstand_i_km / fart_i_kmpt
print(flytid*60)
```

## Klasser, objekter, egenskaper og metoder
I dokumentasjonen for `float` ser vi også at det står `class float()`. `class` eller klasse er et sentralt begrep innen objektorientert programmering. En **klasse** er en mal (oppskrift) på hvordan vi lager et **objekt** av denne klassen.

Tenk på en klasse som en avansert datatype som kan inneholde flere **variabler** og **funksoner**. Variablene kalles for objektets **egenskaper** *(attributes)* og funksjonene kalles for objektets **metoder**. Vi får tak i dem med **punktnotasjon** *(dot notation)*.

```python
class Person:            # Opprett en klasse
    fornavn = ''          # Attributt/egenskap
    etternavn = ''

person = Person()         # Opprett et objekt
person.etternavn = 'Nilsen'          # Punktnotasjon
person.fornavn = 'Kari'
print(person.fornavn, person.etternavn)  # Nilsen
```

Alt i Python er et objekt, og hvert objekt er en instans (forekomst) av minst én klasse. Så fra nå av kan vi slutte å tenke på en variabel som merkelappen på en boks som kan inneholde data og heller innse at en **variabel er en referanse til et objekt**.

La oss gå tilbake til det aller første eksempelet. Vi vet nå at `'Hallo!'` er et tekstobjekt som kan ha metoder som vi kan bruke med punktnotasjon. Når vi skriver et punktum etter `'Hallo!'`, viser IntelliSense egenskaper og metoder som kan benyttes for objekter av klassen `str`. Uten IntelliSense og kjennskap til objektets metoder, ville alternativet vært å lete i str-dokumentasjonen. Vi ser at det finnes en metode `upper` som gjør teksten om til store bokstaver.

Husk å avslutte med metoden med `()`.

```python
print('Hallo!'.upper())  # HALLO!
```
