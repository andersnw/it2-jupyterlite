## Du har brukt objekter hele tiden

I Python er **alt** objekter: tall, tekst, lister og ordbøker. Hver verdi er laget fra en **klasse**, og har **metoder** som vi kaller med punktum. `str`, `list` og `dict` er bare klasser som følger med Python. Når du lager egne klasser, lager du altså nye typer objekter som virker på akkurat samme måte.

Kjør cellen og se:

````python
navn = 'kari'
tall = [3, 1, 2]

print(type(navn))     # navn er et objekt av klassen str
print(type(tall))     # tall er et objekt av klassen list
print(type(42))       # selv tall er objekter

print(navn.upper())   # upper() er en metode i klassen str
tall.append(5)        # append() er en metode i klassen list – den endrer objektet
tall.sort()
print(tall)

tom = list()          # lager et nytt list-objekt – akkurat som Katt() lager et Katt-objekt
print(tom)
````

| | Klasser som følger med Python | Egne klasser (det du skal lære nå) |
|---|---|---|
| Klasse | `str`, `list`, `dict`, `tk.Tk`, `pd.DataFrame` | `Katt`, `Bankkonto`, `Spilleliste` |
| Lage et objekt | `liste = list()`, `vindu = tk.Tk()` | `pus = Katt('Pusur', 3)` |
| Kalle en metode | `navn.upper()`, `df.head()` | `pus.mjau()` |
| Lese et attributt | `df.shape`, `df.columns` (uten parentes!) | `pus.navn` |

Tips: `dir(navn)` viser alle metodene et objekt har. Prøv gjerne!
