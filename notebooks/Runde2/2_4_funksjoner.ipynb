# Funksjoner

Vi har brukt mange funksjoner allerede: innebygde som `print()` og importerte som `mainloop()` i tkinter. I 1.1 laget vi vår egen, `finn_maks()`.

En funksjon er en blokk med kode vi kan kjøre mange ganger ved å **kalle** den. Gjentar vi de samme linjene flere steder, bør de inn i en funksjon. Da blir programmet

* kortere, mer oversiktlig og lettere å lese
* lettere å rette – feil må bare rettes ett sted
* gjenbrukbart – funksjoner kan legges i en modul (`.py`-fil) og importeres i andre programmer (KM07)

Det lønner seg å tenke på funksjoner allerede når vi planlegger: Hvilke trenger vi? Finnes de fra før, eller må vi lage dem?

## 1 Definere og kalle en funksjon

En funksjon defineres med det reserverte ordet `def`, så kommer **funksjonsnavnet** (små bokstaver, beskrivende), parenteser og kolon. Kodelinjene rykkes inn. Funksjonen må være definert før den kalles.

````python
def hils():
    print('Hei, jeg er Eva.')

hils()
````

## 2 Parametere og argumenter

Med **parametere** blir funksjonen mer generell. Parameterne står i parentesen i definisjonen. Verdiene vi sender inn når vi kaller funksjonen, kalles **argumenter**.

````python
def hils(navn):
    print('Hei, jeg er ' + navn + '.')

hils('')
hils('Beate')
````

Glemmer vi argumentet, får vi en feilmelding:

````python
try:
    hils('anders')
except TypeError as feil:
    print('TypeError:', feil)
````

Med flere parametere avgjør **rekkefølgen** hvilket argument som havner hvor – med mindre vi **navngir** argumentene:

````python
def hils(navn,bosted):
    print(f'Hei, jeg er {navn} og kommer fra {bosted}.')

hils('Elise','Ålesund')
hils('Stavanger', 'Frida')              # feil rekkefølge!
hils(bosted='Oslo', navn='Gustav')      # navngitte argumenter
````

### Standardverdier

Gir vi en parameter en **standardverdi**, kan argumentet utelates:

````python
def hils(navn, bosted=None):
    if bosted is None:
        print(f'Hei, jeg er {navn}.')
    else:
        print(f'Hei, jeg er {navn} og kommer fra {bosted}.')

hils('Hans')
hils('Irene', 'Tromsø')
````

### Vilkårlig mange argumenter: `*args` og `**kwargs`

Med `*` foran parameternavnet samles alle posisjonsargumentene i en **tuppel**:

````python
def skriv_ut_fag(*fag):
    if len(fag) == 0:
        print('Jeg liker ingen fag.')
    else:
        tekst = ', '.join(fag)
        før, _, etter = tekst.rpartition(', ') # vs er tuppel. rpartinon returnere alltid tre deler tekst før, skilletegn og tekst etter
        print('Jeg liker ' + (før + ' og ' if før else '') + etter + '.')

skriv_ut_fag()
skriv_ut_fag('engelsk', 'tysk')
skriv_ut_fag('IT', 'biologi', 'psykologi')
````

````python
fag = ('engelsk','IT','matte', 'tysk')
tekst = ', '.join(fag)
print(tekst)
````

* `', '.join(fag)` setter elementene sammen med komma og mellomrom.
* `rpartition(', ')` deler teksten i tre ved det **siste** kommaet (r = fra høyre) og returnerer en tuppel: teksten før, skilletegnet og teksten etter.
* `_` er et vanlig navn på en variabel vi ikke bryr oss om.

Med `**` samles alle **navngitte** argumenter i en **ordbok**:

````python
def skriv_ut_personalia(**personalia):
    if len(personalia) == 0:
        print('Ingen opplysninger.')
    else:
        for nøkkel, verdi in personalia.items():
            print(f'{nøkkel.title():10}: {str(verdi).title()}')
    print()

skriv_ut_personalia()
skriv_ut_personalia(navn='joakim jensen', bosted='jessheim')
skriv_ut_personalia(navn='karl kruse', født=1993, yrke='konsulent')
````

Vanlige parametere må stå **før** `*` og `**`. Parametere som kommer etter `*args`, må navngis når funksjonen kalles.

### Typeannotasjoner (*type hints*)

Python er **dynamisk typet**: en variabel er ikke bundet til én datatype. Vi kan likevel *angi* hvilke typer vi forventer, med **typeannotasjoner**. Python sjekker dem ikke når programmet kjører, men VS Code viser dem når du holder musepekeren over funksjonen, og verktøy som Pylance og mypy kan finne feil.

````python
def areal(bredde: float, høyde: float) -> float:
    return bredde * høyde

print(areal(3, 4.5))
print(areal.__annotations__)
````

## 3 Returverdier

Funksjonene over *gjør* noe. Funksjoner kan også **returnere** en verdi med `return`:

````python
def f(x: float) -> float:
    return x**2 - 4*x + 3

print('f(x) = x² - 4x + 3')
for x in range(6):
    print(f'f({x}) = {f(x)}')
````

Vi kan returnere **flere verdier** – de returneres som en tuppel. Her regner vi ut potenser på to måter, med `**` og med `math.pow`:

````python
import math

def kvadrat_og_kubikk(x: float) -> tuple:
    return x**2, math.pow(x,3)

for x in 1, 3, 6:
    x_2, x_3 = kvadrat_og_kubikk(x)
    print(f'x = {x:2}, x² = {x_2:2} og x³ = {x_3:5}')
````

Eller en **ordbok**:

````python
def sirkelinfo(radius: float) -> dict:
    omkrets = 2 * math.pi * radius
    areal = math.pi * radius**2
    return {'radius': radius, 'omkrets': omkrets, 'areal': areal}

for radius in 3, 4:
    print('Sirkelinfo')
    for nøkkel, verdi in sirkelinfo(radius).items():
        print(f'{nøkkel.title():8}: {verdi:5.2f}')
    print()
````

`math`-modulen har mange funksjoner (trigonometriske, logaritmer, …) og konstantene `pi` og `e`. Skriv `math.` i VS Code, så viser IntelliSense hva som finnes, eller se [dokumentasjonen](https://docs.python.org/3/library/math.html):

## 4 Overføring av argumenter

Parameterne oppfører seg som **lokale variabler**: de finnes bare inne i funksjonen. **Globale variabler** lages utenfor funksjonene og kan leses overalt.

Argumentene overføres som **referanser til objekter**. Hva som skjer, avhenger av om objektet kan endres.

### Uforanderlige objekter (`int`, `float`, `str`, `bool`, `tuple`)

````python
def endre(b: int) -> None:
    print(f'endre, b = {b}')
    b = 8
    print(f'endre, b = {b}')

a = 7
print(f'main,  a = {a}')
endre(a)
print(f'main,  a = {a}')
````

Først peker både `a` og `b` på det samme objektet `7`. Tallet 7 kan ikke endres. `b = 8` lager et *nytt* objekt og lar `b` peke på det – `a` peker fortsatt på 7. Dette virker som **verdioverføring** (*pass by value*). Det spiller ingen rolle om parameteren hadde hett `a`; da hadde vi hatt en lokal og en global `a`.

### Foranderlige objekter (`list`, `dict`, `set`)

````python
def endre(b: list) -> None:
    print(f'endre, b = {b}')
    b.append('Prinsessa')                       # endrer objektet b peker på
    print(f'endre, b = {b}')
    b = ['Per', 'Pål', 'Espen', 'Prinsessa', 'Kongen']   # ny liste
    print(f'endre, b = {b}')

a = ['Per', 'Pål', 'Espen']
print(f'main,  a = {a}')
endre(a)
print(f'main,  a = {a}')
````

`a` og `b` peker på den samme listen. `append` endrer listen, og det ser vi også i `a`. Men når vi lager en ny liste og lar `b` peke på den, peker `a` og `b` ikke lenger på det samme. Derfor kommer Prinsessa med i `a`, men ikke Kongen.

Dette ligner **referanseoverføring** (*pass by reference*). Python kalles gjerne *pass by object reference*: verken det ene eller det andre helt, men litt av begge.

**Tips:** Pythontutor.com visualiserer slike referanser fint: <https://pythontutor.com>

## 5 Lokale og globale variabler

````python
def endre(a: int) -> None:
    print(f'endre: a = {a}, b = {b}')
    a = 5
    globals()['a'] = 6
    print(f'endre: a = {a}, b = {b}')

a = 4
b = 7
print(f'main : a = {a}, b = {b}')
endre(a)
print(f'main : a = {a}, b = {b}')
````

* Inne i funksjonen kan vi **lese** den globale `b`, fordi det ikke finnes en lokal `b`.
* Hadde vi skrevet `b = …` i funksjonen, ville vi laget en lokal `b` som **skygger** for den globale. Det skjer med `a`.
* Den globale `a` når vi med `globals()['a']`, eller ved å skrive `global a` først i funksjonen. `locals()` gir de lokale variablene.

### Rene funksjoner

Vi bør begrense bruken av globale variabler. Her gir den samme funksjonen ulike svar med samme argument, fordi noen har endret den globale `RABATT_GRENSE` i mellomtiden:

````python
RABATT_GRENSE = 1000
RABATT_PROSENT = 20

def rabatt(pris: float) -> float:
    if pris > RABATT_GRENSE:
        return (pris - RABATT_GRENSE) * RABATT_PROSENT / 100
    return 0

pris = 2499
print(f'Pris: {pris:7.2f}, rabatt: {rabatt(pris):7.2f}')
RABATT_GRENSE = 2000
print(f'Pris: {pris:7.2f}, rabatt: {rabatt(pris):7.2f}')
````

Slike funksjoner godtas ikke i matematikken, og heller ikke i funksjonell programmering, der funksjonene skal være **rene**:

* De gir alltid samme svar med samme argumenter.
* De har ingen **bivirkninger** (endrer ikke noe utenfor seg selv).

Det er normalt bedre å *returnere* resultatet enn å lagre det i en global variabel. Store bokstaver (`RABATT_GRENSE`) signaliserer at noe er en **konstant** som ikke skal endres.

````python
def rabatt(pris: float, grense: float = 1000, prosent: float = 20) -> float:
    return max(0, pris - grense) * prosent / 100

print(rabatt(2499), rabatt(2499, grense=2000))
````

### Oppgave 2 – Harmonisk gjennomsnitt
Det er 50 km mellom Ålesund og Ørsta. Det er kø, og vi kjører med snittfart 50 km/t. Hjem igjen er det fritt fram, og snittfarten er 100 km/t. Gjennomsnittsfarten tur/retur er *ikke* 75 km/t (aritmetisk gjennomsnitt), men 67 km/t (**harmonisk gjennomsnitt**) – vi har kjørt 100 km på 1,5 timer.

$$H = \frac{n}{\frac{1}{x_1} + \frac{1}{x_2} + \dots + \frac{1}{x_n}}$$

Lag funksjonen `harm_gjsn(liste)` som returnerer det harmoniske gjennomsnittet.

````python
def harm_gjsn(liste):
    nevner = 0
    for e in liste:
        nevner += 1/e
    return len(liste)/nevner

for liste in [50, 100], [5, 15, 25]:
    print(f'Det harmoniske gjennomsnittet til {liste} er {harm_gjsn(liste):.1f}')
# Forventet: 66.7 og 9.8
````

### Oppgave 3 – Tall med et bestemt siffer
Lag funksjonen `tall_med_siffer(siffer)` som returnerer en liste med alle tall fra 0 til 99 som inneholder sifferet. Skriv ut resultatet med ti tall per linje.

`tall_med_siffer(5)` skal gi `[5, 15, 25, 35, 45, 50, 51, …, 59, 65, 75, 85, 95]`.

````python
# Oppgave 3
tall = list(range(1,100))

def tall_med_siffer(siffer):
    liste = []
    for t in tall:
        if str(siffer) in str(t):
            liste.append(t)
    for i in range(0, len(liste), 10):
        print(*liste[i:i + 10])
        

tall_med_siffer(8)
````
