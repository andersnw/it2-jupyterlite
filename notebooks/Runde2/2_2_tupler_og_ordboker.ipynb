# Tupler og ordbøker

Til slutt bruker vi en **liste med ordbøker** som en enkel tabell, og sorterer og viser den på ulike måter.

## 1 Tupler

En liste (`list`) kan vi endre etter at den er laget: legge til, fjerne og bytte ut elementer.
En **tuppel** ligner på en liste, men er **uforanderlig** (*immutable*). Når den først er laget, ligger innholdet fast.

| | Liste | Tuppel |
|---|---|---|
| Skrives med | `[ ]` | `( )` |
| Kan endres | ja | nei |

Hvorfor bruke tupler?
* De signaliserer at dataene *ikke skal* endres, og beskytter mot utilsiktede feil.
* De bruker litt mindre minne og er litt raskere enn lister.
* De kan brukes som nøkler i ordbøker (det kan ikke lister).

````python
farger = ('rød', 'grønn', 'blå')

print(type(farger))
print(farger)
print('Første farge:', farger[0])
print('Siste farge:', farger[-1])
print('Antall farger:', len(farger))
````

Vi kan gå gjennom en tuppel med en `for`-løkke, akkurat som med en liste:

````python
for farge in farger:
    print(farge.upper(), end=' ')
````

### Tupler kan ikke endres

Prøver vi å bytte ut et element, får vi en `TypeError`. Her fanger vi feilen med `try`/`except` slik at notebooken ikke stopper:

````python
try:
    farger[1] = 'gul'
except TypeError as feil:
    print('Feil:', feil)
````

Trenger vi en endret versjon, må vi lage en **ny** tuppel – eller gjøre den om til en liste:

````python
farger_liste = list(farger)
farger_liste[1] = 'gul'
nye_farger = tuple(farger_liste)

print('Original:', farger)
print('Ny:      ', nye_farger)
````

**Obs!** En tuppel med bare ett element må ha et komma etter elementet:

````python
ikke_tuppel = ('hei')
tuppel = ('hei',)
print(type(ikke_tuppel), type(tuppel))
````

### Pakke ut tupler (*tuple unpacking*)

Parentesene kan faktisk droppes – det er kommaene som lager tuppelen.
Vi kan også **pakke ut** en tuppel til flere variabler i én setning:

````python
punkt = (3, 7)          # samme som (3, 7)
print(type(punkt))

x, y = punkt          # pakker ut
print(f'x = {x}, y = {y}')
````

#### Bytte verdier

Vanligvis trenger vi en hjelpevariabel for å bytte verdiene til to variabler.
Med tupler klarer vi det på én linje:

````python
# Den tradisjonelle måten
a, b = 5, 9
tmp = a
a = b
b = tmp
print(f'a = {a}, b = {b}')

# Med tuppel-utpakking
a, b = b, a
print(f'a = {a}, b = {b}')
````

#### Eksempel: tid i timer, minutter og sekunder

Funksjoner kan returnere flere verdier som en tuppel. Den innebygde funksjonen `divmod(a, b)` gir både heltallsdivisjonen og resten:

````python
print(divmod(17, 5))   # (3, 2) fordi 17 = 3*5 + 2

def timer_min_sek(timer_desimal):
    sekunder = round(timer_desimal * 3600)
    timer, rest = divmod(sekunder, 3600)
    minutter, sekunder = divmod(rest, 60)
    return timer, minutter, sekunder      # returnerer en tuppel

t, m, s = timer_min_sek(2.37)
print(f'{t} timer, {m} minutter og {s} sekunder')
````

Det samme kan gjøres med `datetime.timedelta`. Teksten `'2:22:12'` deles opp med `split(':')`, og delene pakkes rett ut i tre variabler:

````python
import datetime

varighet = datetime.timedelta(hours=2.37)
print(varighet)

tt, mm, ss = str(varighet).split(':')
print(f'Timer: {tt}, minutter: {mm}, sekunder: {ss}')
````

## 2 Ordbøker

I en liste finner vi elementene ved hjelp av et **tall** (indeks).
I en **ordbok** (*dictionary*, datatypen `dict`) bruker vi i stedet en **nøkkel**, gjerne et navn.
Hvert element er et **nøkkel/verdi-par** (*key-value pair*). Nøklene må være unike.

Ordbøker lages med krøllparenteser `{ }`, og vi henter, endrer, legger til og sletter verdier med nøkkelen i hakeparenteser.

````python
elev = {'navn': 'Ingrid', 'klasse': '2STA'}

print(type(elev))
print(elev)
print(elev['navn'])          # les en verdi

elev['klasse'] = '3STA'      # endre en verdi
elev['poeng'] = 87           # legg til et nytt par
print(elev)

del elev['klasse']           # slett et par
print(elev)
````

### Nøkler som ikke finnes

Spør vi etter en nøkkel som ikke finnes, får vi `KeyError`. Med metoden `get` kan vi i stedet få en standardverdi:

````python
print(elev.get('klasse'))               # None
print(elev.get('klasse', 'ukjent'))     # 'ukjent'
print('poeng' in elev)                  # sjekk om nøkkelen finnes
````

### Løkker over ordbøker

Vi kan gå gjennom bare nøklene, bare verdiene eller begge deler:

| Metode | Gir |
|---|---|
| `.keys()` | nøklene |
| `.values()` | verdiene |
| `.items()` | par av (nøkkel, verdi) som tupler |

````python
bok = {
    'tittel': 'Sult',
    'forfatter': 'Knut Hamsun',
    'utgitt': 1890,
    'sjanger': 'roman',
}

for nøkkel in bok.keys():
    print(nøkkel, end=' ')
print()

for verdi in bok.values():
    print(verdi, end=' ')
print()

for nøkkel, verdi in bok.items():       # tuppel-utpakking igjen!
    print(f'{nøkkel.capitalize()}: {verdi}')
````

## 3 En liste med ordbøker – en enkel tabell

Vi kan samle flere ordbøker i en liste. Tenk på hver ordbok som en **rad** og nøklene som **kolonneoverskrifter**.

````python
elever = [
    {'navn': 'Ingrid', 'poeng': 87},
    {'navn': 'Sander', 'poeng': 64},
    {'navn': 'Amalie', 'poeng': 92},
    {'navn': 'Jonas',  'poeng': 75},
]
````

### Største og minste verdi

`max` og `min` trenger å vite *hva* de skal sammenligne. Det oppgir vi med parameteren `key`.
`itemgetter('poeng')` lager en funksjon som henter verdien til nøkkelen `'poeng'` fra hver ordbok.

````python
from operator import itemgetter

best = max(elever, key=itemgetter('poeng'))
svakest = min(elever, key=itemgetter('poeng'))

print(f"{best['navn']} har flest poeng ({best['poeng']}).")
print(f"{svakest['navn']} har færrest poeng ({svakest['poeng']}).")
````

En `lambda`-funksjon gjør nøyaktig det samme, og er nyttig å kjenne til:

````python
best = max(elever, key=lambda e: e['poeng'])
print(best)
````

### Sortering

`sorted` lager en **ny** sortert liste og lar originalen være urørt. Med `reverse=True` sorteres det synkende.

````python
print('Alfabetisk etter navn')
for e in sorted(elever, key=itemgetter('navn')):
    print(f"{e['navn']:<8}{e['poeng']:>4} p")

print('\nEtter poeng, høyest først')
for e in sorted(elever, key=itemgetter('poeng'), reverse=True):
    print(f"{e['navn']:<8}{e['poeng']:>4} p")

print('\nOriginal liste (uendret)')
#print(elever)
print('\n'.join(str(e) for e in elever))
````

**Formatering i f-strenger:** `:<8` venstrejusterer i 8 tegns bredde, `:>4` høyrejusterer i 4 tegns bredde. Det gir pene kolonner.

Vil vi sortere selve listen (og ikke lage en ny), bruker vi metoden `elever.sort(key=...)`.

## 5 Liste eller ordbok?

Når vi sletter et element fra en liste, flytter alle elementene etter det seg én plass fram og får ny indeks.
Det kan gi overraskende feil hvis vi sletter mens vi går gjennom listen forfra:

````python
tall = [4, 7, 8, 10, 3]

# Forsøk: slett alle partall ved å gå forfra
for i in range(len(tall)):
    if i < len(tall) and tall[i] % 2 == 0:
        del tall[i]
print('Forfra: ', tall)    # 8 ble hoppet over!

tall = [4, 7, 8, 10, 3]
# Gå bakfra – da påvirker ikke slettingen indeksene vi har igjen
for i in range(len(tall) - 1, -1, -1):
    if tall[i] % 2 == 0:
        del tall[i]
print('Bakfra:', tall)
````

**Tenk etter:** Hvorfor ble 8 hoppet over i første forsøk?

I en ordbok skjer ikke dette, fordi elementene finnes via nøkkelen og ikke via posisjonen.
Oppslag med nøkkel er dessuten svært raskt, og koden blir lettere å lese: `elev['poeng']` sier mer enn `elev[1]`.
Fra og med Python 3.7 husker ordbøker også rekkefølgen elementene ble lagt inn i.

## 6 Oversikt over sammensatte datatyper

| | `list` | `tuple` | `dict` | `set` |
|---|---|---|---|---|
| Skrives med | `[1, 2]` | `(1, 2)` | `{'a': 1}` | `{1, 2}` |
| Rekkefølge | ordnet | ordnet | ordnet (fra 3.7) | uordnet |
| Kan endres | ja | **nei** | ja | ja |
| Duplikater | tillatt | tillatt | **ikke** like nøkler | **ikke** like verdier |
| Tilgang via | indeks | indeks | nøkkel | – (kun medlemskap) |

**Obs!** `{}` lager en tom *ordbok*. En tom mengde lages med `set()`.

````python
print(type({}), type(set()))
print(set([3, 1, 3, 2, 1]))   # duplikater fjernes
````


**Tips:** En ordbok kan koble bokstaven til nøkkelen det skal sorteres etter, f.eks. `{'T': 'tittel', ...}`.
I Jupyter kan du tømme utskriften med `clear_output()` fra `IPython.display`. I et vanlig terminalprogram brukes `os.system('cls')` på Windows og `os.system('clear')` på Mac/Linux.

````python
from IPython.display import clear_output

bøker = [
    {'tittel': 'Sult',              'forfatter': 'Hamsun',   'år': 1890},
    {'tittel': 'Et dukkehjem',      'forfatter': 'Ibsen',    'år': 1879},
    {'tittel': 'Kristin Lavransdatter', 'forfatter': 'Undset', 'år': 1920},
    {'tittel': 'Fuglane',           'forfatter': 'Vesaas',   'år': 1957},
    {'tittel': 'Is-slottet',        'forfatter': 'Vesaas',   'år': 1963},
]

# Kobler bokstaven brukeren skriver til nøkkelen vi sorterer etter
valg_til_nøkkel = {'T': 'tittel', 'F': 'forfatter', 'Å': 'år'}

while True:
    print('Sorter etter: T = tittel, F = forfatter, Å = år, S = slutt')
    valg = input('Ditt valg: ').upper()
    clear_output()

    if valg == 'S':
        print('Takk for nå!')
        break                   # Hopper ut av løkka, istedet for stoppkriterie etter while (nå er det bare True)

    try:
        nøkkel = valg_til_nøkkel[valg]      # gir KeyError ved ugyldig valg
    except KeyError:
        print('Ukjent valg, prøv igjen.\n')
        continue     # kjør løkka på nytt

    print(f"{'Tittel':<24}{'Forfatter':<12}{'År':>6}")
    print('-' * 42)
    for bok in sorted(bøker, key=lambda x: x[nøkkel]):
        print(f"{bok['tittel']:<24}{bok['forfatter']:<12}{bok['år']:>6}")
    print()
````

### Oppgave 4 – Grafisk brukergrensesnitt

Lag et program med samme funksjonalitet som i oppgave 3, men med grafisk brukergrensesnitt i tkinter.
For eksempel kan du ha én knapp for hver sorteringsmåte, og vise tabellen med `Label`-er i et rutenett eller i en `ttk.Treeview`.

**Utvidelse:** Klikk på samme knapp to ganger for å bytte mellom stigende og synkende sortering.

````python
# Oppgave 4
from tkinter import Tk, Label, Button

bøker = [
    {'tittel': 'Sult',              'forfatter': 'Hamsun',   'år': 1890},
    {'tittel': 'Et dukkehjem',      'forfatter': 'Ibsen',    'år': 1879},
    {'tittel': 'Kristin Lavransdatter', 'forfatter': 'Undset', 'år': 1920},
    {'tittel': 'Fuglane',           'forfatter': 'Vesaas',   'år': 1957},
    {'tittel': 'Is-slottet',        'forfatter': 'Vesaas',   'år': 1963},
]

# Husker forrige sortering, slik at samme knapp to ganger snur rekkefølgen
forrige = {'nøkkel': '', 'synkende': False}


def lag_tabelltekst(liste):
    # Samme f-strenger som i oppgave 3, men samlet i én tekst
    tekst = f"{'Tittel':<24}{'Forfatter':<12}{'År':>6}\n"
    tekst += '-' * 42 + '\n'
    for bok in liste:
        tekst += f"{bok['tittel']:<24}{bok['forfatter']:<12}{bok['år']:>6}\n"
    return tekst


def sorter(nøkkel):
    if nøkkel == forrige['nøkkel']:
        forrige['synkende'] = not forrige['synkende']   # snu rekkefølgen
    else:
        forrige['synkende'] = False                     # ny knapp: stigende
    forrige['nøkkel'] = nøkkel

    sortert = sorted(bøker, key=lambda x: x[nøkkel], reverse=forrige['synkende'])
    tabell.config(text=lag_tabelltekst(sortert))


def sorter_tittel():
    sorter('tittel')

def sorter_forfatter():
    sorter('forfatter')

def sorter_år():
    sorter('år')


vindu = Tk()
vindu.title('Bøker')

Button(vindu, text='Tittel', width=12, command=sorter_tittel).grid(row=0, column=0)
Button(vindu, text='Forfatter', width=12, command=sorter_forfatter).grid(row=0, column=1)
Button(vindu, text='År', width=12, command=sorter_år).grid(row=0, column=2)

# Courier er en skrift der alle tegn er like brede, så kolonnene blir rette
tabell = Label(vindu, text=lag_tabelltekst(bøker), font=('Courier', 12), justify='left')
tabell.grid(row=1, column=0, columnspan=3, padx=10, pady=10)

vindu.mainloop()
````

### Oppgave 5 – Tenk og forklar

a) Hvorfor kan en tuppel brukes som nøkkel i en ordbok, mens en liste ikke kan det? Prøv begge deler.

b) Gi et eksempel på data som passer best i en tuppel, og et eksempel som passer best i en ordbok.

#### Svar på oppgave 5

**a)** En ordbok finner nøklene sine raskt ved å regne ut et tall fra nøkkelen, en *hash-verdi*. Tallet forteller hvor i ordboken verdien ligger. Da må nøkkelen gi samme tall hver gang, og derfor kan den ikke endres.

* En **tuppel** kan ikke endres, så den kan brukes som nøkkel.
* En **liste** kan endres. Hvis lista ble endret etter at den var lagt inn, ville ordboken lete på feil sted. Derfor nekter Python å bruke den som nøkkel.

```python
avstander = {('Oslo', 'Bergen'): 463}     # tuppel som nøkkel: OK
print(avstander[('Oslo', 'Bergen')])      # 463

avstander[['Oslo', 'Trondheim']] = 494    # liste som nøkkel
# TypeError: unhashable type: 'list'
```

**b)**

* **Tuppel:** data med et fast antall deler, der hver plass har en bestemt betydning og innholdet ikke skal endres. Eksempler er et koordinat `(59.91, 10.75)`, en RGB-farge `(255, 0, 0)` og en dato `(2026, 10, 27)`.
* **Ordbok:** data der vi vil slå opp med et navn i stedet for en posisjon, eller der innholdet kan endres. Eksempler er en elev `{'navn': 'Ingrid', 'klasse': '2STA', 'poeng': 87}` og en telefonliste `{'Ingrid': '91234567', 'Sander': '98765432'}`.
